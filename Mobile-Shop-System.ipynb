```python
import pandas as pd

# Product data
product_data = {
    'Item_Code': ['BPCM','BPSH','RPSS','RPLL','YPLS','YPLL',
                  'RTMS','RTML','YTLM','YTLL','SMNO','SMPG',
                  'CSST','CSLX','CGCR','CGHM'],
    'Description': ['Compact', 'Clam shell', 'Robo phone - 5in 64GB',
                    'Robo phone - 6in 256GB', 'Y-phone 6in 64GB',
                    'Y-phone 6in 256GB', 'RoboTab 8in 64GB',
                    'RoboTab 10in 128GB', 'Y-tab 10in 128GB',
                    'Y-tab 10in 256GB', 'SIM Free', 'Pay As You Go',
                    'Case Standard', 'Case Luxury',
                    'Car Charger', 'Home Charger'],
    'Price': [29.99, 49.99, 199.99, 499.99, 549.99, 649.99,
              149.99, 299.99, 499.99, 599.99, 0.00, 9.99,
              0.00, 50.00, 19.99, 15.99]
}

products_df = pd.DataFrame(product_data)
print(products_df)
```

# TASK 1: FIRST DEVICE PURCHASE

```python
def get_price(code):
    return float(products_df.loc[products_df['Item_Code'] == code, 'Price'].values[0])

def get_description(code):
    return products_df.loc[products_df['Item_Code'] == code, 'Description'].values[0]

# Variables to store first purchase
items_bought = []
total_price = 0

print("\n=== TASK 1: FIRST DEVICE PURCHASE ===")

# Device selection
while True:
    device_type = input("Choose device (phone/tablet): ").lower()
    if device_type in ['phone','tablet']:
        break
    print("Invalid input. Type phone or tablet.")

# Phone or Tablet selection
if device_type == "phone":
    phone_codes = ['BPCM','BPSH','RPSS','RPLL','YPLS','YPLL']
    print("\nAvailable Phones:")
    for code in phone_codes:
        print(code, "-", get_description(code), "-", get_price(code))

    while True:
        phone_choice = input("Enter phone code: ").upper()
        if phone_choice in phone_codes:
            break
        print("Invalid code.")

    items_bought.append(phone_choice)
    total_price += get_price(phone_choice)


    

    # SIM
    print("SIM options: SMNO (0), SMPG (9.99)")
    while True:
        sim_choice = input("Choose SIM (SMNO/SMPG): ").upper()
        if sim_choice in ['SMNO','SMPG']:
            break
        print("Invalid SIM code.")

    items_bought.append(sim_choice)
    total_price += get_price(sim_choice)

else:
    tablet_codes = ['RTMS','RTML','YTLM','YTLL']
    print("\nAvailable Tablets:")
    for code in tablet_codes:
        print(code, "-", get_description(code), "-", get_price(code))

    while True:
        tab_choice = input("Enter tablet code: ").upper()
        if tab_choice in tablet_codes:
            break
        print("Invalid tablet code.")

    items_bought.append(tab_choice)
    total_price += get_price(tab_choice)



# Case


print("\nCase options: CSST (0), CSLX (50)")
while True:
    case_choice = input("Choose case (CSST/CSLX): ").upper()
    if case_choice in ['CSST','CSLX']:
        break
    print("Invalid case code.")

items_bought.append(case_choice)
total_price += get_price(case_choice)




# Chargers



if input("Add Car Charger? (yes/no): ").lower() == "yes":
    items_bought.append("CGCR")
    total_price += get_price("CGCR")
if input("Add Home Charger? (yes/no): ").lower() == "yes":
    items_bought.append("CGHM")
    total_price += get_price("CGHM")


# Summary of first device


print("\n--- TASK 1 SUMMARY ---")
for code in items_bought:
    print(code, "-", get_description(code), "-", get_price(code))
print("First device total: €", round(total_price,2))
```

# TASK 2: ADDITIONAL DEVICES.  

```python

additional_devices = []
additional_total = 0

device_count = 0  # counts additional devices

while True:
    more = input("\nDo you want to add another device? (yes/no): ").lower()
    if more != "yes":
        break

    device_count += 1
    print(f"\n=== ADDITIONAL DEVICE #{device_count} ===")

    new_items = []
    new_price = 0

    # Device selection
    while True:
        device_type = input("Choose device (phone/tablet): ").lower()
        if device_type in ['phone','tablet']:
            break
        print("Invalid input. Type phone or tablet.")

    if device_type == "phone":
        phone_codes = ['BPCM','BPSH','RPSS','RPLL','YPLS','YPLL']
        print("\nAvailable Phones:")
        for code in phone_codes:
            print(code, "-", get_description(code), "-", get_price(code))

        while True:
            phone_choice = input("Enter phone code: ").upper()
            if phone_choice in phone_codes:
                break
            print("Invalid code.")

        new_items.append(phone_choice)
        new_price += get_price(phone_choice)

        

        # SIM


        
        print("SIM options: SMNO (0), SMPG (9.99)")
        while True:
            sim_choice = input("Choose SIM (SMNO/SMPG): ").upper()
            if sim_choice in ['SMNO','SMPG']:
                break
            print("Invalid SIM code.")

        new_items.append(sim_choice)
        new_price += get_price(sim_choice)

    else:
        tablet_codes = ['RTMS','RTML','YTLM','YTLL']
        print("\nAvailable Tablets:")
        for code in tablet_codes:
            print(code, "-", get_description(code), "-", get_price(code))

        while True:
            tab_choice = input("Enter tablet code: ").upper()
            if tab_choice in tablet_codes:
                break
            print("Invalid tablet code.")

        new_items.append(tab_choice)
        new_price += get_price(tab_choice)

    

    # Case

    
    print("\nCase options: CSST (0), CSLX (50)")
    while True:
        case_choice = input("Choose case (CSST/CSLX): ").upper()
        if case_choice in ['CSST','CSLX']:
            break
        print("Invalid case code.")

    new_items.append(case_choice)
    new_price += get_price(case_choice)

    # Chargers
    if input("Add Car Charger? (yes/no): ").lower() == "yes":
        new_items.append("CGCR")
        new_price += get_price("CGCR")
    if input("Add Home Charger? (yes/no): ").lower() == "yes":
        new_items.append("CGHM")
        new_price += get_price("CGHM")

    additional_devices.extend(new_items)
    additional_total += new_price

    
    # Device summary

    
    print("\n--- ADDITIONAL DEVICE SUMMARY ---")
    for code in new_items:
        print(code, "-", get_description(code), "-", get_price(code))
    print(f"Device total: €{round(new_price,2)}")
```

# TASK 3: DISCOUNT FOR ADDITIONAL DEVICES

```python

discount_total = additional_total * 0.10  # 10% off additional devices
final_additional_total = additional_total - discount_total

grand_total = total_price + final_additional_total

# FINAL SUMMARY
print("\n==============================")
print("          FINAL ORDER")
print("==============================")

# First device
print("\n--- FIRST DEVICE ---")
for code in items_bought:
    print(code, "-", get_description(code), "-", get_price(code))
print("First device total: €", round(total_price,2))

# Additional devices
if additional_devices:
    print("\n--- ADDITIONAL DEVICES ---")
    for code in additional_devices:
        print(code, "-", get_description(code), "-", get_price(code))
    print(f"Total for additional devices before discount: €{round(additional_total,2)}")

    print("\n==============================")
    print("  ---- Discount----")
    print("==============================")    
    print(f"Total discount: €{round(discount_total,2)}")
    print(f"Total for additional devices after discount: €{round(final_additional_total,2)}")

print("\n=== GRAND TOTAL TO PAY: €", round(grand_total,2), "===")
```
